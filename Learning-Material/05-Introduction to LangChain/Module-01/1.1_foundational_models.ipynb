## Initializing and Invoking a Model

```python
!pip install langchain-groq
```

```python
from langchain.chat_models import init_chat_model
```

```python
from groq import Groq
from google.colab import userdata

GROQ_API_KEY = userdata.get('groq_key')
```

```python
model = init_chat_model(model_provider="groq",
                        model= "llama-3.3-70b-versatile",
                        groq_api_key=GROQ_API_KEY)
```

```python
response = model.invoke("What is the capital of Moon?")
response
```

```python
response.content
```

```python
from pprint import pprint
pprint(response.response_metadata)
```

## Initializing and Invoking an Agent

```python
from langchain.agents import create_agent
```

```python
agent= create_agent(model=model)
```

```python
from langchain.messages import HumanMessage
```

```python
response = agent.invoke(
    {"messages": [HumanMessage(content="What is the capital of Moon?")]}
)
```

```python
pprint(response)
```

```python
print(response['messages'][-1].content)
```

```python
from langchain.messages import AIMessage
```

```python
response = agent.invoke(
    {"messages": [HumanMessage(content="What's the capital of the Moon?"),
    AIMessage(content="The capital of the Moon is Luna City."),
    HumanMessage(content="Interesting, tell me more about Luna City")]}
)
```

```python
pprint(response)
```

##Streaming Mode

```python
for token, metadata in agent.stream(
    {"messages": [HumanMessage(content="Tell me all about Luna City, the capital of the Moon")]},
    stream_mode="messages"
):

    # token is a message chunk with token content
    # metadata contains which node produced the token

    if token.content:  # Check if there's actual content
        print(token.content, end="", flush=True)  # Print token
```

```python
system_prompt= "You are a science fiction writer, create a capital city at user's request"
scifi_agent= create_agent(model=model, system_prompt=system_prompt)
response = scifi_agent.invoke({"messages": [HumanMessage(content="What is the capital of the Moon?")]})
print(response['messages'][-1].content)
```

```python
system_prompt= """
You are a science fiction writer, create a capital city at user's request. You are required to respond as the examples given below:
User: what is the capital of Mars?
SciFi Writer: Marsialis.

User: what is the capital of Venus?
SciFi Writer: Venusovia
"""
scifi_agent= create_agent(model=model, system_prompt=system_prompt)
response = scifi_agent.invoke({"messages": [HumanMessage(content="What is the capital of the Moon?")]})
print(response['messages'][-1].content)
```

## Structured Output

```python
from pydantic import BaseModel
```

```python
class Capital_Info(BaseModel):
  name:str
  location: str
  vibe: str
  economy: str
```

```python
agent= create_agent(model=model,
                    system_prompt= "You are a science fiction writer, create a capital city at user's request",
                    response_format= Capital_Info)
response = agent.invoke({"messages": [HumanMessage(content="What is the capital of the Moon?")]})
response['structured_response']
```
